```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy as sp
from itertools import combinations 
import ast
from sklearn.linear_model import LogisticRegression
import seaborn as sn
%matplotlib inline

data_folder = './data/'
```

### A) Load the dataset and convert categorical features to a suitable numerical representation (use dummy-variable encoding). 
- Split the data into a training set (80%) and a test set (20%). Pair each feature vector with the corresponding label, i.e., whether the outcome_type is adoption or not. 
- Standardize the values of each feature in the data to have mean 0 and variance 1.

The use of external libraries is not permitted in part A, except for numpy and pandas. 
You can drop entries with missing values.

```python
columns = ['animal_type', 'intake_year', 'intake_condition', 'intake_number', 'intake_type', 'sex_upon_intake', \
          'age_upon_intake_(years)', 'time_in_shelter_days', 'sex_upon_outcome', 'age_upon_outcome_(years)', \
          'outcome_type']
original_data = pd.read_csv(data_folder+'aac_intakes_outcomes.csv', usecols=columns)
```

```python
print('The length of the data with all rows is : {}'.format(len(original_data)))
original_data.dropna(inplace=True)
# dropna() - removes rows or columns with missing/null values
# inplace=True - modifies the DataFrame directly instead of returning a new one
print('The length of the data without the rows with nan value is: {}'.format(len(original_data)))
```

```python
data_features = original_data.copy()
data_features['adopted'] = data_features.outcome_type.apply(lambda r: 1 if r=='Adoption' else 0) # Create a new binary column 'adopted': 1 if outcome_type is 'Adoption', else 0
# .apply(...) runs a function on each row value of the outcome_type column.
# lambda r: 1 if r=='Adoption' else 0 maps the string 'Adoption' to 1 and any other outcome to 0.
data_features.drop("outcome_type", axis = 1, inplace=True) # Drop the original 'outcome_type' column as it's no longer needed
#  axis = 1 indicates that we are dropping a column (not a row).
data_features.head()
```

```python
def split_set(data_to_split, ratio=0.8):
    mask = np.random.rand(len(data_to_split)) < ratio # It creates a random boolean mask of the same length as data_to_split. Each position is True with probability equal to ratio and False otherwise.
    return [data_to_split[mask].reset_index(drop=True), data_to_split[~mask].reset_index(drop=True)]
# reset_index(drop=True) - resets the index of the resulting DataFrames after filtering, dropping the old index.
```

```python
[train, test] = split_set(data_features)
```

```python
categorical_columns = ['sex_upon_outcome', 'animal_type', 'intake_condition',
                       'intake_type', 'sex_upon_intake']
train_categorical = pd.get_dummies(train, columns=categorical_columns)
train_categorical.columns
```

The features in the testing set must be matched with the traning set.

```python
# Make sure we use only the features available in the training set
test_categorical = pd.get_dummies(test, columns=categorical_columns)[train_categorical.columns] # Align test set columns with training set columns
# The indexing [train_categorical.columns] ensures the test DataFrame has exactly the same columns in the same order as training.
```

```python
train_label=train_categorical.adopted # Target variable for training set
train_features = train_categorical.drop('adopted', axis=1)
print('Length of the train dataset : {}'.format(len(train)))

test_label=test_categorical.adopted
test_features = test_categorical.drop('adopted', axis=1)
print('Length of the test dataset : {}'.format(len(test)))
```

**Z-score standardization** transforms features to have mean = 0 and standard deviation = 1 using the formula:

$$z = \frac{x - \mu}{\sigma}$$

Where:
- $x$ = original value
- $\mu$ = mean of the feature
- $\sigma$ = standard deviation of the feature

This ensures all features are on the same scale, which is important for algorithms like logistic regression.

```python
means = train_features.mean() # Compute the mean of each feature in the training set
stddevs = train_features.std() # Compute the standard deviation of each feature in the training set

# Standardize the training set
train_features_std = pd.DataFrame()
for c in train_features.columns:
    train_features_std[c] = (train_features[c]-means[c])/stddevs[c]

# Use the mean and stddev of the training set (moi: feature scaling must be consistent between train and test.)
test_features_std = pd.DataFrame()
for c in test_features.columns:
    test_features_std[c] = (test_features[c]-means[c])/stddevs[c]

train_features_std.head()
```

## B) Train a logistic regression classifier on your training set. Logistic regression returns probabilities as predictions, so in order to arrive at a binary prediction, you need to put a threshold on the predicted probabilities. 
- For the decision threshold of 0.5, present the performance of your classifier on the test set by displaying the confusion matrix. Based on the confusion matrix, manually calculate accuracy, precision, recall, and F1-score with respect to the positive and the negative class. 

```python
def compute_confusion_matrix(true_label, prediction_proba, decision_threshold=0.5): 
    
    predict_label = (prediction_proba[:,1]>decision_threshold).astype(int) # Convert probabilities to binary predictions based on the threshold
    # prediction_proba[:, 1] selects the probability of the positive (adopted = 1) class for each sample.
    # decision_threshold compares each probability to the chosen threshold (e.g., 0.5), producing booleans (True if above threshold).
    # .astype(int) turns True/False into 1/0, giving the predicted class labels.                                     

    TP = np.sum(np.logical_and(predict_label==1, true_label==1)) # np.logical_and(...) keeps only rows where both are true, and np.sum(...) counts them
    TN = np.sum(np.logical_and(predict_label==0, true_label==0))
    FP = np.sum(np.logical_and(predict_label==1, true_label==0))
    FN = np.sum(np.logical_and(predict_label==0, true_label==1))
    
    confusion_matrix = np.asarray([[TP, FP],
                                    [FN, TN]])
    return confusion_matrix


def plot_confusion_matrix(confusion_matrix):
    [[TP, FP],[FN, TN]] = confusion_matrix
    label = np.asarray([['TP {}'.format(TP), 'FP {}'.format(FP)],
                        ['FN {}'.format(FN), 'TN {}'.format(TN)]])
    
    df_cm = pd.DataFrame(confusion_matrix, index=['Yes', 'No'], columns=['Positive', 'Negative']) # Passing confusion_matrix into pd.DataFrame with the given index and columns simply wraps those counts in a labeled table for plotting
    
    return sn.heatmap(df_cm, cmap='YlOrRd', annot=label, annot_kws={"size": 16}, cbar=False, fmt='')


def compute_all_score(confusion_matrix, t=0.5):
    [[TP, FP],[FN, TN]] = confusion_matrix.astype(float)
    
    accuracy =  (TP+TN)/np.sum(confusion_matrix)
    
    precision_positive = TP/(TP+FP) if (TP+FP) !=0 else np.nan
    precision_negative = TN/(TN+FN) if (TN+FN) !=0 else np.nan
    
    recall_positive = TP/(TP+FN) if (TP+FN) !=0 else np.nan
    recall_negative = TN/(TN+FP) if (TN+FP) !=0 else np.nan

    F1_score_positive = 2 *(precision_positive*recall_positive)/(precision_positive+recall_positive) if (precision_positive+recall_positive) !=0 else np.nan
    F1_score_negative = 2 *(precision_negative*recall_negative)/(precision_negative+recall_negative) if (precision_negative+recall_negative) !=0 else np.nan

    return [t, accuracy, precision_positive, recall_positive, F1_score_positive, precision_negative, recall_negative, F1_score_negative]
```

```python
logistic = LogisticRegression(solver='lbfgs', max_iter=10000)
logistic.fit(train_features_std,train_label)
```

```python
prediction_proba = logistic.predict_proba(test_features_std)
```

```python
confusion_matrix_05 = compute_confusion_matrix(test_label, prediction_proba, 0.5 )
plt.figure(figsize = (4,3)) 
ax = plot_confusion_matrix(confusion_matrix_05)
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.title('Confusion matrix for a 0.5 threshold')
```

```python
[t, accuracy, precision_positive, recall_positive, F1_score_positive, \
    precision_negative, recall_negative, F1_score_negative] = compute_all_score(confusion_matrix_05)

print("The accuracy of this model is {0:1.3f}".format(accuracy)) # {0:1.3f} - formats it as a fixed-point float with 3 digits after the decimal (.3f), using at least 1 character before the decimal (1)
print("For the positive case, the precision is {0:1.3f}, the recall is {1:1.3f} and the F1 score is {2:1.3f}"\
      .format(precision_positive, recall_positive, F1_score_positive))
print("For the negative case, the precision is {0:1.3f}, the recall is {1:1.3f} and the F1 score is {2:1.3f}"\
      .format(precision_negative, recall_negative, F1_score_negative))
```

## (moi) Interprétation des coefficients

Ces coefficients représentent l'impact de chaque feature sur la probabilité d'adoption:

**Valeur (magnitude):** Plus le coefficient est grand en valeur absolue, plus cette feature influence fortement la prédiction.

**Signe:**
- **Coefficient positif** → augmente la probabilité d'adoption (pousse vers "adopté")
- **Coefficient négatif** → diminue la probabilité d'adoption (pousse vers "pas adopté")

### Forme mathématique

La prédiction suit d'abord une **forme linéaire**:

$$z = \text{intercept} + \text{coef}_1 \times \text{feature}_1 + \text{coef}_2 \times \text{feature}_2 + \ldots$$

Puis la **régression logistique applique une fonction sigmoid**:

$$P(\text{adoption} = 1) = \frac{1}{1 + e^{-z}}$$

Les coefficients affichés correspondent à la partie linéaire, mais leur contribution est transformée par le sigmoid pour produire une probabilité entre 0 et 1.

## Questions 1: Which of the following metrics is most suitable when you are dealing with unbalanced classes?

- **a) F1 Score**
- b) Recall
- c) Precision
- d) Accuracy

a)

### (moi)

**Problem with accuracy:** If 95% of animals are adopted, a model that predicts "adopted" for everything gets 95% accuracy but is useless. Accuracy is heavily skewed by the dominant class.

**F1 Score balances both classes:**

$$F1 = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$$

It penalizes the model if it performs poorly on either the minority or majority class, making it ideal for imbalanced datasets.<br>
F1 Score ensures good performance on both classes (adopted or not), which matters most when data is imbalanced.

## Question 2: You are working on a binary classification problem. You trained a model on a training dataset and got the following confusion matrix on the test dataset. What is true about the evaluation metrics (rounded to the second decimal point):

|            | Pred = NO|Pred=YES|
|------------|----------|--------|
| Actual NO  |    50    |   10   |
| Actual YES |    5     |   100  |

- a) Accuracy is 0.95
- b) Accuracy is 0.85
- c) False positive rate is 0.95
- **d) True positive rate is 0.95**

d)

### (moi) True Positive Rate Calculation

**True Positive Rate (TPR)** (also called **Recall** or **Sensitivity**):

$$TPR = \frac{TP}{TP + FN}$$

**Where:**
- **TP (True Positives)** = 100 (correctly predicted YES)
- **FN (False Negatives)** = 5 (actual YES but predicted NO)

**Calculation:**

$$TPR = \frac{100}{100 + 5} = \frac{100}{105} = 0.952 \approx 0.95$$

**Interpretation:** Of all animals that were actually adopted (105 total), the model correctly identified 95% of them.
