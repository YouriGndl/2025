````python
import random
import math
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import make_blobs, make_moons
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
from sklearn.manifold import TSNE
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

%matplotlib inline
````

# Part A) Tutorial

In this tutorial, you will learn how to discover clusters in data using Python.

Let's start by creating some **[synthetic](https://en.wikipedia.org/wiki/Synthetic_data)** data!
First, we create a super secret number that represents the number of the cluster to generate. 

Don't print it! We'll try to discover it automatically.

## K-Means

````python
random.seed(42)
top_secret_number = random.randint(2, 10) # Randomly chosen number of clusters between 2 and 10
````

Now we can generate some data distributed in n groups:

````python
total_samples = 100

# This create some artifical clusters with standard dev. = 2
X, _, centers = make_blobs(n_samples=total_samples, 
                           centers=top_secret_number, 
                           cluster_std=2,
                           n_features=2, # 2D data
                           return_centers=True, # we also want to get the centers
                           random_state=42)

X[:3]
````

Let's see how these clusters look like and where is their center:

````python
plt.scatter(X[:,0], X[:,1], alpha=0.6)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Artificial clusters (%s samples)" % total_samples) # %s is string formatting

for c in centers:
    plt.scatter(c[0], c[1], marker="+", color="red")
````

### Clustering the data
How many clusters do you see? Probably easy for a human, but not so trivial for a computer. 

Let's try to group the data with K-Means. Recall that K-Means requires you to specify the number of clusters (K). Let's start by testing multiple values between 2 and 10:

````python
MIN_CLUSTERS = 2
MAX_CLUSTERS = 10

# Compute number of row and columns
COLUMNS = 3
ROWS = math.ceil((MAX_CLUSTERS-MIN_CLUSTERS)/COLUMNS)
fig, axs = plt.subplots(ROWS, COLUMNS, figsize=(10,8), sharey=True, sharex=True)

# Plot the clusters
for n_clusters in range(MIN_CLUSTERS, MAX_CLUSTERS+1):
    current_column = (n_clusters-MIN_CLUSTERS)%COLUMNS # % is the modulo operator
    current_row = (n_clusters-MIN_CLUSTERS)//COLUMNS # // is the integer division operator
    # Get the axis where to add the plot
    ax = axs[current_row, current_column]
    # Cluster the data with the current number of clusters
    kmean = KMeans(n_clusters=n_clusters, random_state=42).fit(X)
    # Plot the data by using the labels as color
    ax.scatter(X[:,0], X[:,1], c=kmean.labels_, alpha=0.6) # c=kmean.labels_ colors the points according to their cluster
    ax.set_title("%s clusters"%n_clusters)
    ax.set_xlabel("Feature 1")
    ax.set_ylabel("Feature 2")
    # Plot the centroids
    for c in kmean.cluster_centers_:
        ax.scatter(c[0], c[1], marker="+", color="red")

# kmean.labels_
    # array([0, 1, 2, 0, 1, 2, ...])  # Length = 100 (one label per sample)

# kmean.cluster_centers_
    # array([[ 5.2,  3.1],
    #        [ 2.8, -1.5],
    #        [-4.1,  2.9]])  # 3 centers × 2 features


plt.tight_layout()
````

### Bonus example: 
Did you know you can animate your plots?

````python
from matplotlib.animation import FuncAnimation
fig, ax = plt.subplots()


X_3clusters, _, centers = make_blobs(n_samples=100, 
                           centers=3, 
                           cluster_std=2,
                           return_centers=True, 
                           random_state=99)

# Update frame callback 
def update(max_iter):
    # Clean the plot
    plt.clf() # clf: clear the entire figure
    plt.cla() # cla: clear the current axis
    # Cluster and get the labels
    kmeans = KMeans(n_clusters=3, 
                    init="random", 
                    algorithm="lloyd",
                    random_state=10, 
                    n_init=1, max_iter=max_iter).fit(X_3clusters) # n_init=1 parameter in KMeans controls how many times the algorithm runs with different random initializations.
    # Plot
    plt.scatter(X_3clusters[:,0], X_3clusters[:,1], c=kmeans.labels_, alpha=0.6)
    for c in kmeans.cluster_centers_:
        plt.scatter(c[0], c[1], marker="+", color="red")
    plt.title("%s iterations" % max_iter)

# Animate 1 frame per second
anim = FuncAnimation(fig, update, frames=np.arange(1, 20), interval=500)

# Save the gif
anim.save('clusters.gif', dpi=80) # dpi=80 for better quality
````

### How to select K in K-Means?

You have a couple of options:
- Silhouette score: Find the K with the desired tradeoff between the number of clusters and cohesion/separation.
- Elbow method: Find the "elbow" in the curve of the Sum of Squared Errors

Let's see what we can observe using these 2 methods.

#### Silhouette score

To generate the curve of the silhouette score, we cluster the data with different value of K and plot the resulting values.

````python
silhouettes = []

# Try multiple k
for k in range(2, 11):
    # Cluster the data and assigne the labels
    labels = KMeans(n_clusters=k, random_state=10).fit_predict(X) # fit_predict because just need the labels
    # Get the Silhouette score
    score = silhouette_score(X, labels)
    silhouettes.append({"k": k, "score": score})
    
# Convert to dataframe
silhouettes = pd.DataFrame(silhouettes)

# Plot the data
plt.plot(silhouettes.k, silhouettes.score)
plt.xlabel("K")
plt.ylabel("Silhouette score")
````

#### Elbow method

Similarly to the previous case, we compute the SSE for different values of K, and we plot the distribution:

````python
def plot_sse(features_X, start=2, end=11):
    sse = []
    for k in range(start, end):
        # Assign the labels to the clusters
        kmeans = KMeans(n_clusters=k, random_state=10).fit(features_X)
        sse.append({"k": k, "sse": kmeans.inertia_}) # kmeans.inertia_ is the sum of squared distances (sse (error)) from each point to its nearest centroid. It's also called the Within-Cluster Sum of Squares (WCSS).

    sse = pd.DataFrame(sse)
    # Plot the data
    plt.plot(sse.k, sse.sse)
    plt.xlabel("K")
    plt.ylabel("Sum of Squared Errors")
    
plot_sse(X)
````

In this case, we get the same recommendation from both methods: K=3. Silhouette suggests that using 3 clusters is a fair tradeoff between the number of groups and their separation. The elbow method shows how the SSE reduction is less significant with more than 3 clusters.

Let's plot the best clusters:


````python
fig, axs = plt.subplots(1, 1, figsize=(4,4), sharey=True)

# Plot the clusters with K = 3
labels = KMeans(n_clusters=3, random_state=0).fit_predict(X)
axs.scatter(X[:,0], X[:,1], c=labels, alpha=0.6)
````

Ultimately it is up to you (domain expert!) to pick the number of clusters that better represent the data.

The best representation, in this case, is obtained with k = 3.

### Visualizing high dimensional data

Visualizing 2-dimensional data is easy, but what happens when we have more than three features? 

We can use dimensionality reduction techniques! Let start by generating some artifical clusters with 10 features. 
We use the same number of samples and groups as the previous example:

````python
total_samples = 100

# This create some artifical clusters with standard dev. = 3
X10d, _ = make_blobs(n_samples=total_samples, 
                           centers=top_secret_number, 
                           cluster_std=3,
                           n_features=10,
                           random_state=0)

print("The features of the first sample are: %s" % X10d[0])
````

To transform this features vector in a form that can be represented in a simple plot, we can reduce the number of dimensions by preserving as much information as possible. Let's see two techniques that can help in this task:

- t-SNE
- PCA


Note: the next lecture will cover more theoretical aspects of these methods.

#### t-SNE

````python
X_reduced_tsne = TSNE(n_components=2, init='random', learning_rate='auto', random_state=0).fit_transform(X10d)
# n_components=2 because we want to reduce to 2 dimensions for visualization
# init='random': Initialization method for embedding
# learning_rate='auto': Controls optimization step size
# random_state=0: Seed for reproducibility
print("The features of the first sample are: %s" % X_reduced_tsne[0])
````

(moi): Embedding refers to a representation of data in a lower-dimensional space.

````python
X_reduced_pca = PCA(n_components=2).fit(X10d).transform(X10d) # same as .fit_transform(X10d)

print("The features of the first sample are: %s" % X_reduced_pca[0])
````

The features are reduced in both cases to a 2d space. Please note that they are not the same because the two techniques optimize different objectives.

````python
fig, axs = plt.subplots(1, 2, figsize=(7,3), sharey=True)

# Cluster the data in 3 groups
labels = KMeans(n_clusters=3, random_state=0).fit_predict(X10d) # K-Means is performed on the original 10-dimensional data (X10d), not on the 2D reduced version.

# Plot the data reduced in 2d space with t-SNE
axs[0].scatter(X_reduced_tsne[:,0], X_reduced_tsne[:,1], c=labels, alpha=0.6) # [:,0] is NumPy array slicing that selects all rows from the first column (column 0).
axs[0].set_title("t-SNE")

# Plot the data reduced in 2d space with PCA
axs[1].scatter(X_reduced_pca[:,0], X_reduced_pca[:,1], c=labels, alpha=0.6)
axs[1].set_title("PCA")
````

### Density based: DBSCAN

K-Means is an intuitive and effective algorithm, but what happens when the data has a complex shape?

As for the previous example, let's generate some synthetic data that resemble two half-moons.

````python
# Create 500 random points with 2-moons shape
X_moons, _ = make_moons(500, noise=0.05, random_state=0)

plt.scatter(X_moons[:,0], X_moons[:,1], alpha=0.6)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
````

We can visually see that there are two groups. Let's assign the labels with K-Means!

````python
kmean = KMeans(n_clusters=2, random_state=0).fit(X_moons)
plt.scatter(X_moons[:,0], X_moons[:,1], c=kmean.labels_, alpha=0.6)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

# Plot the centroids
for c in kmean.cluster_centers_:
    plt.scatter(c[0], c[1], marker="o", color="red")
````

Let's cluster the data with DBSCAN by variating the `eps` value in the range between 0.05 and 0.15.

eps: is the epsilon parameter in DBSCAN — it defines the maximum distance between two points for them to be considered neighbors.

````python
# Create a list of eps
eps_list = np.linspace(0.05, 0.15, 14)

# Compute number of row and columns
COLUMNS = 7
ROWS = math.ceil(len(eps_list)/COLUMNS)

fig, axs = plt.subplots(ROWS, COLUMNS, figsize=(12, 4), sharey=True, sharex=True)

for i in range(0, len(eps_list)):
    eps = eps_list[i]
    
    current_column = i%COLUMNS
    current_row = i//COLUMNS
    
    ax = axs[current_row, current_column]
    labels = DBSCAN(eps=eps).fit_predict(X_moons)
    ax.scatter(X_moons[:,0], X_moons[:,1], c=labels, alpha=0.6)
    ax.set_title("eps = {:.3f}".format(eps))
    
plt.tight_layout()
````

## (moi) Unsupervised Learning Workflow

1. **Determine optimal K** (number of clusters)
   - Use Elbow method or Silhouette score
   - Example: Find K=3 is optimal

2. **Reduce dimensionality with PCA** (optional but recommended)
   - Reduce from original dimensions to fewer dimensions
   - Example: 13 features → 5 dimensions
   - Helps with curse of dimensionality

3. **Cluster with K-Means** on reduced data
   - Apply K-Means in the reduced dimensional space
   - Example: KMeans(n_clusters=3) on 5D data
   - Get cluster labels for each sample

4. **Visualize with t-SNE** (for human interpretation)
   - Reduce to 2D for visualization only
   - Example: 13 features → 2D t-SNE plot
   - Color points by K-Means labels
   - Do NOT cluster on t-SNE data

1. Netflix matrix factorization pipeline to discover users with similar interests
2. **Speaker recognition (recognition of the identity of who is talking) in phones and smart assistant devices**
3. LDA (moi: Latent Dirichlet Allocation.) topic modeling on Twitter content to discover customers' opinions about a product

**(moi) Explanation:**

**Speaker recognition is supervised learning**, not unsupervised learning.

**Why:**
- **Requires labeled training data**: The system needs examples of each person's voice labeled with their identity (e.g., "This audio is John," "This audio is Mary")
- **Classification task**: Given new audio, predict which person is speaking
- **Ground truth needed**: You must know who's talking to train the model

**The other options are unsupervised:**
1. Matrix factorization — discovers user patterns without labels
3. Topic modeling — finds topics in text without predefined categories
4. K-means clustering — groups domains without knowing true categories

**Key distinction:** Supervised = labeled data + prediction task. Unsupervised = no labels + discover patterns.

### Question 2: Which of the following is true for the k-means clustering algorithm?

1. It always converges to a global optimum.
2. **In each iteration, points are assigned to the closest centroid and the new centroids are recomputed.**
3. It outputs the optimal number of clusters.
4. None of the above.

2)

**(moi) Explanation:**

This statement accurately describes **K-Means' iterative algorithm**:

1. **Assignment step**: Each point is assigned to its nearest centroid (closest by Euclidean distance)
2. **Update step**: Centroids are recomputed as the mean of all points assigned to that cluster
3. **Repeat** until convergence (centroids stop moving)

**Why the other options are wrong:**

- **Option 1** (global optimum): ❌ K-Means converges to a **local optimum**, which depends on initial centroid positions
- **Option 3** (optimal K): ❌ You must specify K beforehand; K-Means doesn't determine the optimal number of clusters
- **Option 4**: ❌ Incorrect since option 2 is true

## Key Takeaways

### 1. **Clustering is an Exploratory Tool**
Unsupervised learning allows us to discover patterns without prior knowledge. The wheat seeds and wine examples showed how clustering can automatically identify groups that correspond to real categories (wheat varieties, wine cultivars) without being told they exist.

### 2. **Algorithm Selection Matters**
- **K-Means**: Fast and effective for spherical, well-separated clusters; requires specifying K
- **DBSCAN**: Better for irregular shapes and varying densities; requires tuning epsilon parameter
- Choose based on your data's geometry and characteristics

### 3. **Optimal K Determination is Critical**
Both Elbow method and Silhouette score provide guidance, but domain expertise is essential for final decisions. The "right" number of clusters depends on the problem context, not just mathematical optimization.

### 4. **Preprocessing is Essential**
StandardScaler (z-score normalization) ensures all features contribute equally to distance calculations. Distance-based algorithms like K-Means are highly sensitive to feature scales.

### 5. **Dimensionality Reduction Serves Two Purposes**
- **For clustering**: PCA can reduce computational cost and mitigate curse of dimensionality
- **For visualization**: t-SNE and PCA enable human interpretation of high-dimensional results
- **Critical distinction**: Always cluster in the appropriate feature space (moi: dimension), then visualize

### 6. **Practical Workflow**
```
1. Preprocess (scale features)
2. Determine optimal K (Elbow/Silhouette)
3. Optional: Reduce dimensionality with PCA
4. Cluster in feature space (K-Means/DBSCAN)
5. Visualize results in 2D (t-SNE/PCA)
```
